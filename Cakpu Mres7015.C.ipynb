```python
import zipfile, shutil
from pathlib import Path

zip_path = Path("/content/NLP_Package.zip")
assert zip_path.is_file(), "Re-run Cell 2 and upload NLP_Package.zip"

for name in ["care_notes_nlp_framework", "Synthetic_CareNotes_Dataset"]:
    p = Path("/content") / name
    if p.exists():
        shutil.rmtree(p)

with zipfile.ZipFile(zip_path, "r") as z:
    z.extractall("/content")

fw = Path("/content/care_notes_nlp_framework")
ds = Path("/content/Synthetic_CareNotes_Dataset")
print("Framework:", fw.is_dir())
print("Dataset:", ds.is_dir())
print("Corpus:", (ds / "care_notes_corpus.csv").is_file())
assert fw.is_dir() and (ds / "care_notes_corpus.csv").is_file()
print("Ready.")
```

```python
from pathlib import Path
from google.colab import files
import zipfile

results = Path("/content/care_notes_nlp_framework/results")
out_zip = Path("/content/NLP_Results.zip")
with zipfile.ZipFile(out_zip, "w", zipfile.ZIP_DEFLATED) as z:
    for f in results.rglob("*"):
        if f.is_file() and f.suffix.lower() in {".json", ".csv", ".png", ".txt"}:
            z.write(f, arcname=str(f.relative_to("/content")))
print("Downloading", out_zip)
files.download(str(out_zip))
```
